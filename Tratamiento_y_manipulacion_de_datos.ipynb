# **PROBLEMA DE NEGOCIO**

Se trata de un algoritmo capaz de analizar las caracteristicas de una propiedad como comodidades, tamaño, ocupación del espacio en un periodo determinado y sugerir al anfitrion un precio a cobrar por tarifas diarias que garantice ganancias en momentos de alta demanda.


##**Entendiendo el problema**

```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_json("/content/datos_hosting.json")
df
```

```python
df= pd.json_normalize(df["info_inmuebles"])
df
```

```python
columnas= list(df.columns)
columnas
```

```python
df= df.explode(columnas[3:])
df
```

```python
df.reset_index(inplace=True, drop=True)
df
```

##**Datos Numericos**

```python
df.info()
```

```python
columnas_numericas= ["max_hospedes", "cantidad_baños", "cantidad_cuartos", "cantidad_camas"]
```

```python
df[columnas_numericas]= df[columnas_numericas].astype(np.int64)
df.info()
```

```python
df["evaluacion_general"]= df["evaluacion_general"].astype(np.float64)
df.info()
```

```python
columnas_decimales= ["cuota_deposito", "cuota_limpieza", "precio"]
```

```python
df[columnas_decimales]= df[columnas_decimales].applymap(lambda x: x.replace("$", "").replace(",","").strip())
df
```

```python
df.info()
```

```python
df[columnas_decimales]= df[columnas_decimales].astype(np.float64)
df.info()
```

##**Datos de Texto**

```python
df.head()
```

```python
df["descripcion_local"]= df["descripcion_local"].str.lower()
df["descripcion_local"]
```

```python
df["descripcion_local"][3169]
```

```python
df["descripcion_local"]= df["descripcion_local"].str.replace("[^a-zA-Z0-9\-\']"," ",regex=True)
df["descripcion_local"]= df["descripcion_local"].str.replace("(?<!\w)-(?!\w)"," ",regex=True)
df["descripcion_local"]= df["descripcion_local"].str.split()
df["descripcion_local"]
```

```python
df["comodidades"]= df["comodidades"].str.replace('\{|}|\"'," ",regex=True)
df["comodidades"]= df["comodidades"].str.split(",")
df["comodidades"]
```
