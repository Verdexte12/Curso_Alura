```python
import pandas as pd
df= pd.read_csv("/content/base_tiempo.csv")
df.sample(5)
```

##**Anotaciones y otros graficos**

```python
fig= plt.figure(figsize=(12,5))
eje= fig.add_axes([0,0,1,1])
eje.grid(True)
eje.plot(df["date"], df["temperatura"], color="g")
eje.set_title("Temperatura en el tiempo", size=25, pad=20)
eje.legend(["temperatura"], loc="lower right", fontsize=15)
eje.set_xlabel("Año", size=12)
eje.set_ylabel("Temperatura", size=12)
eje.axhline(df["temperatura"].max(), color="r", linestyle="--")
eje.axhline(df["temperatura"].min(), color="r", linestyle="--")

plt.show()
```

```python
fig= plt.figure(figsize=(12,5))
eje= fig.add_axes([0,0,1,1])
eje.grid(True)
eje.plot(df["date"], df["temperatura"], color="g")
eje.set_title("Temperatura en el tiempo", size=25, pad=20)
eje.legend(["temperatura"], loc="lower right", fontsize=15)
eje.set_xlabel("Año", size=12)
eje.set_ylabel("Temperatura", size=12)
eje.axhline(df["temperatura"].max(), color="r", linestyle="--")
eje.axhline(df["temperatura"].min(), color="r", linestyle="--")
x1= df["date"][df["temperatura"].idxmax()]
y1= df["temperatura"].max()
x2= df["date"][df["temperatura"].idxmin()]
y2= df["temperatura"].min()
eje.annotate("Temperatura maxima", xy=(x1,y1), fontsize=12, color="k")
eje.annotate("Temperatura minima", xy=(x2,y2), fontsize=12, color="k")

plt.show()
```

```python
fig= plt.figure(figsize=(12,5))
eje= fig.add_axes([0,0,1,1])
eje.grid(True)
eje.plot(df["date"], df["temperatura"], color="g")
eje.set_title("Temperatura en el tiempo", size=25, pad=20)
eje.legend(["temperatura"], loc="lower right", fontsize=15)
eje.set_xlabel("Año", size=12)
eje.set_ylabel("Temperatura", size=12)
eje.axhline(df["temperatura"].max(), color="r", linestyle="--")
eje.axhline(df["temperatura"].min(), color="r", linestyle="--")
x1= df["date"][df["temperatura"].idxmax()]
y1= df["temperatura"].max()
x1_2= df["date"][df["temperatura"].idxmax() - 8000]
y1_2= df["temperatura"].max() - 5
x2= df["date"][df["temperatura"].idxmin()]
y2= df["temperatura"].min()
x2_2= df["date"][df["temperatura"].idxmin() - 8000]
y2_2= df["temperatura"].min() + 5
eje.annotate("Temperatura maxima", xy=(x1,y1), fontsize=12, color="k", xytext=(x1_2,y1_2) ,arrowprops=dict(facecolor="black", shrink=0.05))
eje.annotate("Temperatura minima", xy=(x2,y2), fontsize=12, color="k", xytext=(x2_2,y2_2) ,arrowprops=dict(facecolor="black", shrink=0.05))

plt.show()
```

```python
temperatura_por_dia_de_semana= df.groupby("dia_de_semana")["temperatura"].mean()
nombre_dias= ["Domingo","Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado"]
temperatura_por_dia_de_semana= temperatura_por_dia_de_semana[nombre_dias]
temperatura_por_dia_de_semana
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
indice= range(len(temperatura_por_dia_de_semana))
eje.bar(indice, temperatura_por_dia_de_semana)
eje.set_title("Temperatura por dia de la semana", size=15)
eje.set_xlabel("Dia de la semana", size=12)
eje.set_ylabel("Temperatura promedio", size=12)
eje.set_xticks(indice)
eje.set_xticklabels(temperatura_por_dia_de_semana.index)
plt.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
indice= range(len(temperatura_por_dia_de_semana))
colores=["r","g","b","y","c","m","k"]

eje.bar(indice, temperatura_por_dia_de_semana, color=colores)
eje.set_title("Temperatura por dia de la semana", size=15)
eje.set_xlabel("Dia de la semana", size=12)
eje.set_ylabel("Temperatura promedio", size=12)
eje.set_xticks(indice)
eje.set_xticklabels(temperatura_por_dia_de_semana.index)
plt.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
indice= range(len(temperatura_por_dia_de_semana))
colores=["r","g","b","y","c","m","k"]

eje.bar(indice, temperatura_por_dia_de_semana, color=colores, edgecolor="black")
eje.set_title("Temperatura por dia de la semana", size=15)
eje.set_xlabel("Dia de la semana", size=12)
eje.set_ylabel("Temperatura promedio", size=12)
eje.set_xticks(indice)
eje.set_xticklabels(temperatura_por_dia_de_semana.index)
plt.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
eje.pie(temperatura_por_dia_de_semana, labels=temperatura_por_dia_de_semana.index, autopct="%1.1f%%")
eje.set_title("Temperatura por dia de la semana", size=15)
plt.show()
```

#**Aula 4**

##**Graficos de dispersion**

```python
df= pd.read_csv("/content/iris.csv")
df
```

```python
fig= plt.figure(figsize=(15,8))
eje= fig.add_axes([0,0,1,1])

colores= {"Iris-setosa":"r", "Iris-versicolor":"b", "Iris-virginica":"g"}
marcadores= {"Iris-setosa":"x", "Iris-versicolor":"o", "Iris-virginica":"v"}
for especie in df["Species"].unique():
  tmp= df[df["Species"] == especie]
  eje.scatter(tmp["SepalLengthCm"], tmp["SepalWidthCm"], color=colores[especie], marker=marcadores[especie], s=100)

eje.set_title("Grafico de dispersion", size=25, pad=15)
eje.set_xlabel("SepalLengthCm", size=12)
eje.set_ylabel("SepalWidthCm", size=12)
eje.tick_params(labelsize=15)
eje.tick_params(labelsize=15)
eje.legend(colores.keys())
fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
eje.boxplot(df.drop("Species", axis=1).values)
eje.set_title("Grafico de caja", size=15, pad=10)
eje.set_xticklabels(df.drop("Species", axis=1).columns)
fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
eje.boxplot(df.drop("Species", axis=1).values, patch_artist=True)
eje.set_title("Grafico de caja", size=15, pad=10)
eje.set_xticklabels(df.drop("Species", axis=1).columns)
fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
colores= ["red","blue","orange","green"]
cajas= eje.boxplot(df.drop("Species", axis=1).values, patch_artist=True)
for caja, color in zip(cajas["boxes"], colores):
  caja.set_facecolor(color)
for outlier in cajas["fliers"]:
  outlier.set(marker="x")
eje.set_title("Grafico de caja", size=15, pad=10)
eje.set_xticklabels(df.drop("Species", axis=1).columns)
fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])

eje.hist(df["PetalLengthCm"])
eje.set_title("Histograma", size=15, pad=10)
eje.set_xlabel("PetalLengthCm", size=12)
eje.set_ylabel("Frecuencia", size=12)
eje.grid(True)
fig.show()
```

##**Personalizando histogramas**

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
mu,sigma=df["PetalLengthCm"].mean(), df["PetalLengthCm"].std()
eje.hist(df["PetalLengthCm"], bins=20)
eje.set_title("Histograma", size=15, pad=10)
eje.set_xlabel("PetalLengthCm", size=12)
eje.set_ylabel("Frecuencia", size=12)
eje.annotate("$mu = {0:.2f}$\n$sigma = {1:.2f}$".format(mu,sigma), xy=(4.5,20), fontsize=20)
eje.grid(True)

fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
mu,sigma=df["PetalLengthCm"].mean(), df["PetalLengthCm"].std()
eje.hist(df["PetalLengthCm"], bins=20)
eje.set_title("Histograma", size=15, pad=10)
eje.set_xlabel("PetalLengthCm", size=12)
eje.set_ylabel("Frecuencia", size=12)
eje.annotate("$\mu = {0:.2f}$\n$\sigma = {1:.2f}$".format(mu,sigma), xy=(4.5,20), fontsize=20)
eje.grid(True)
eje.axvline(mu, color="r", linestyle="--")
eje.axvline(df['PetalLengthCm'].median(), color="r", linestyle="--")
eje.annotate("promedio", xy=(mu-1.3,28), fontsize=15)
eje.annotate("mediana", xy=(df['PetalLengthCm'].median(),31), fontsize=15)


fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
df_iv= df[df["Species"] == "Iris-versicolor"]
mu,sigma=df_iv["PetalLengthCm"].mean(), df_iv["PetalLengthCm"].std()
eje.hist(df_iv["PetalLengthCm"], bins=20)
eje.set_title("Iris versicolor", size=15, pad=10)
eje.set_xlabel("PetalLengthCm", size=12)
eje.annotate("$\mu = {0:.2f}$\n$\sigma = {1:.2f}$".format(mu,sigma), xy=(4.6,6), fontsize=20)
eje.grid(True)
eje.axvline(mu, color="r", linestyle="--")
eje.axvline(df_iv['PetalLengthCm'].median(), color="g", linestyle="--")
eje.annotate("promedio", xy=(mu-0.5,5.5), fontsize=15, color="r")
eje.annotate("mediana", xy=(df_iv['PetalLengthCm'].median()-0.4,6.5), fontsize=15, color="g")
fig.savefig("histograma_iv.png", bbox_inches="tight")

fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
df_is= df[df["Species"] == "Iris-setosa"]
mu,sigma=df_is["PetalLengthCm"].mean(), df_is["PetalLengthCm"].std()
eje.hist(df_is["PetalLengthCm"], bins=20)
eje.set_title("Iris setosa", size=15, pad=10)
eje.set_xlabel("PetalLengthCm", size=12)
eje.annotate("$\mu = {0:.2f}$\n$\sigma = {1:.2f}$".format(mu,sigma), xy=(1.7,7), fontsize=20)
eje.grid(True)
eje.axvline(mu, color="r", linestyle="--")
eje.axvline(df_is['PetalLengthCm'].median(), color="g", linestyle="--")
eje.annotate("promedio", xy=(mu-0.2,13.5), fontsize=15, color="r")
eje.annotate("mediana", xy=(df_is['PetalLengthCm'].median()+0.05,12), fontsize=15, color="g")
fig.savefig("histograma_is.png", bbox_inches="tight")

fig.show()
```

```python
fig= plt.figure(figsize=(5,4))
eje= fig.add_axes([0,0,1,1])
df_ivc= df[df["Species"] == "Iris-virginica"]
mu,sigma=df_ivc["PetalLengthCm"].mean(), df_ivc["PetalLengthCm"].std()
eje.hist(df_ivc["PetalLengthCm"], bins=20)
eje.set_title("Iris virginica", size=15, pad=10)
eje.set_xlabel("PetalLengthCm", size=12)
eje.annotate("$\mu = {0:.2f}$\n$\sigma = {1:.2f}$".format(mu,sigma), xy=(6.3,7), fontsize=20)
eje.grid(True)
eje.axvline(mu, color="r", linestyle="--")
eje.axvline(df_ivc['PetalLengthCm'].median(), color="g", linestyle="--")
eje.annotate("promedio", xy=(mu,8), fontsize=15, color="r")
eje.annotate("mediana", xy=(df_ivc['PetalLengthCm'].median(),7), fontsize=15, color="g")
fig.savefig("histograma_ivg.png", bbox_inches="tight")

fig.show()
```
