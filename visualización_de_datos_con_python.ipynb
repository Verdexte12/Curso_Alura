#**Visualizacion de datos con Python: Matplotlib, Seaborn, Plotly**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df= pd.read_csv("/content/inmigrantes_canada.csv")
df
```

```python
df.info()
```

```python
df.set_index("Pais", inplace=True)
df
```

```python
anos= list(map(str,range(1980,2014)))
anos
```

```python
colombia= df.loc["Colombia",anos]
colombia
```

```python
col_dict= {"Año":colombia.index.tolist(),
           "Inmigrantes":colombia.values.tolist()}
col_dict
```

```python
df_colombia= pd.DataFrame(col_dict)
df_colombia
```

```python
plt.figure(figsize=(8,4))
plt.plot(df_colombia["Año"], df_colombia["Inmigrantes"])
plt.xticks(["1980","1985","1990","1995","2000","2005","2010"])
plt.title("Inmigrantes de Colombia")
ax = plt.gca()
ax.set_xlabel("Años")
ax.set_ylabel("Inmigrantes")
plt.show()
```

```python
dif, ax = plt.subplots(figsize=(8,4))
ax.plot(df_colombia["Año"], df_colombia["Inmigrantes"])
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title("Inmigrantes de Colombia")
ax.set_xlabel("Años")
ax.set_ylabel("Inmigrantes")
plt.show()
```

```python
fig, axs= plt.subplots(1,2, figsize=(8,4))
axs[0].plot(df_colombia["Año"], df_colombia["Inmigrantes"])
axs[0].set_title("Inmigrantes de Colombia")
axs[0].set_xlabel("Años")
axs[0].set_ylabel("Inmigrantes")
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid(True)

axs[1].boxplot(df_colombia["Inmigrantes"])
axs[1].set_title("BoxPlot de Inmigrantes de Colombia")
axs[1].set_xlabel("Colombia")
axs[1].set_ylabel("Inmigrantes")
axs[1].grid(True)
plt.show()
```

```python
fig, axs= plt.subplots(2,2, figsize=(10,6))
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle("Inmigrantes por País")

axs[0,0].plot(df.loc["Colombia", anos])
axs[0,0].set_title("Colombia")

axs[0,1].plot(df.loc["Brasil", anos])
axs[0,1].set_title("Brasil")

axs[1,0].plot(df.loc["Argentina", anos])
axs[1,0].set_title("Argentina")

axs[1,1].plot(df.loc["Perú", anos])
axs[1,1].set_title("Perú")

ymin= 0
ymax= 7000

for ax in axs.ravel():
  ax.set_ylim(ymin, ymax)

for ax in axs.flat:
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  ax.set_xlabel("Años")
  ax.set_ylabel("Inmigrantes")
  ax.grid(True)

plt.show()
```

```python
from matplotlib.lines import lineStyles
dif, ax = plt.subplots(figsize=(8,4))
ax.plot(df_colombia["Año"], df_colombia["Inmigrantes"],lw=2,color="green")#marker="o")
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title("Inmigrantes de Colombia", fontsize=18)
ax.set_xlabel("Años", fontsize=14)
ax.set_ylabel("Inmigrantes", fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
ax.grid(linestyle= "--")
plt.show()
```

```python
sudamerica_sorted= sudamerica.sort_values("Total", ascending=True)
sudamerica_sorted
colores= ["Royalblue", "orange", "forestgreen", "orchid", "purple", "brown",
          "slateblue", "gray", "olive", "navy", "teal", "tomato"]
fig, ax = plt.subplots(figsize=(12,5))

ax.barh(sudamerica_sorted.index, sudamerica_sorted["Total"],color=colores)
ax.set_title("Inmigrantes por País en América del Sur", loc="left", fontsize=18)
ax.set_ylabel("", fontsize=14)
ax.set_xlabel("Inmigrantes", fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
plt.show()
```

```python
colores=[]
for pais in sudamerica_sorted.index:
  if pais== "Colombia":
    colores.append("forestgreen")
  else:
    colores.append("silver")

fig, ax = plt.subplots(figsize=(12,5))

ax.set_frame_on(False)
ax.xaxis.set_visible(False)
ax.tick_params(axis="y", which="both", size=0)

ax.barh(sudamerica_sorted.index, sudamerica_sorted["Total"],color=colores)
ax.set_title("Inmigrantes de Colombia", loc="left", fontsize=18)
ax.set_ylabel("", fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
#ax.set_xlabel("Inmigrantes", fontsize=14)
#ax.xaxis.set_tick_params(labelsize=12)

for i,j in enumerate(sudamerica_sorted["Total"]):
  ax.text(j+20,i,str(j),color="black", fontsize=8.6,ha="left", va="center")

plt.show()
```

```python
top_10= df.sort_values("Total", ascending= False).head(10)
top_10
```

```python
ax= sns.barplot(data=top_10,x="Total", y=top_10.index,orient="h")
ax.set(title="Inmigrantes por País", xlabel="Inmigrantes", ylabel="")
ax.bar_label(ax.containers[0])

plt.show()
```

```python
fig, ax= plt.subplots(figsize=(8,4))

ax= sns.barplot(data=top_10,x="Total", y=top_10.index,orient="h")
ax.set_title("Inmigrantes por País",loc="left",fontsize=16)
ax.set_xlabel("Inmigrantes",fontsize=14)
ax.set_ylabel("")
ax.bar_label(ax.containers[0])
plt.show()
```

```python
def generar_grafico(palette):
  fig, ax= plt.subplots(figsize=(8,4))
  ax= sns.barplot(data=top_10,x="Total", y=top_10.index, hue=top_10.index,orient="h", palette=palette, legend=False)
  ax.set_title("Inmigrantes por País",loc="left",fontsize=16)
  ax.set_xlabel("Inmigrantes",fontsize=14)
  ax.set_ylabel("")
  ax.set(xticklabels=[])
  sns.despine()
  #ax.set_frame_on(False)
  #ax.xaxis.set_visible(False)
  #ax.tick_params(axis="y", which="both", size=0)

  for i,j in enumerate(top_10["Total"]):
    ax.text(j+20,i,str(j),color="black", fontsize=8.6,ha="left", va="center")

  plt.show()
```

```python
sns.set_theme(style="darkgrid")
generar_grafico("tab10")
```

```python
sns.set_theme(style="whitegrid")
generar_grafico("tab10")
```
